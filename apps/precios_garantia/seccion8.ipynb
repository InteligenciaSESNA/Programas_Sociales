```python
import pandas as pd
import numpy as np
import plotly as plt
import plotly.graph_objects as go
import plotly.io as pio
```

```python
df = pd.read_excel('../../datasets/beneficiarios_mun.xlsx', converters={'cve_ent':str, 'cve_mun':str})
df.head()
```

```python
df.columns
```

```python
# Filtros
anio = 2020
producto = 'Frijol'
entidad = 'Zacatecas'

df_filt = df.copy()
# filtramos
df_filt = df_filt[df_filt['year'] == anio]
df_filt = df_filt[df_filt['cultivo'] == producto]
df_filt = df_filt[df_filt['entidad'] == entidad]
df_filt = df_filt.sort_values('monto_total',ascending=False)
df_filt.head()
```

```python


fig = go.Figure()
fig.add_trace(go.Bar(x=df_filt['municipio'].to_list(), y=df_filt['monto_total'].to_list()))

fig.update_layout(
        showlegend=False,
        autosize=True,
        #width=650,
        height=400,
        margin=dict(
            l=0,
            r=0,
            b=0,
            t=60,
            pad=0),
            plot_bgcolor='white',
            paper_bgcolor="white",
            )

fig.update_layout(
        title="Monto de apoyos por producto",
        xaxis_title="Entidad",
        yaxis_title="Monto del Apoyo ($)",
        legend_title="",
        font=dict(
            #family="Courier New, monospace",
            size=12,
            color="#2a3240"
            ))


fig
```
